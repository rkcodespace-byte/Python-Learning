```python
# Factorial
# Create a function to calculate the factorial of a number.
# Example: factorial(5) → 120


def fact(f):
    count = 1
    for i in range(1,f+1):
        count = count * i
    return count


        
x= int(input("Enter the Factorial Number:"))
p=fact(x)

print(p)
```

```python
def vowels(words):
    count=0

    for i in words:
        if i in "aeiou":
            count = count + 1

    return count

text= input("Enter your words:")
c= vowels(text)

print("Total Vowels:",c)
```

```python
# Reverse a String
# Create a function that reverses a given string.
# Example: reverse("Python") → "nohtyP"

def reverse(word):
    return word[::-1]

text = input("Enter a word: ")
result = reverse(text)

print("Reversed String:", result)

        
```

```python
def grade(marks):

    if marks >= 90:
        return "A"

    elif marks >= 80:
        return "B"

    elif marks >= 70:
        return "C"

    elif marks >= 60:
        return "D"

    else:
        return "F"


num = int(input("Enter your marks: "))

if 0 <= num <= 100:
    print("Your Grade:", grade(num))
else:
    print("Invalid Marks")
    
```

```python
# Task: Write a Python program using a for loop to:
# 1. Print numbers from 1 to 10.
# 2. Print only even numbers from 1 to 10.
# 3. Calculate the sum of numbers from 1 to 10.

print("Numbers:")
for i in range (1,11):
    print(i)

print("Even Numbers:")
for j in range(2,11,2):
    print(j)

count = 0
for k in range (1,11):
    count += k

print("Sum:",count)
```
